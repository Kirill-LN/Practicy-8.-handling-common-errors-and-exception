# 1) Создайте собственное исключение для проверки, что введенное пользователем число является положительным. Если число отрицательное, выбросьте ваше исключение.

```python
class NegativeNumber(Exception):
    """ Пользовательноское исключение """
    pass

def check_number(number: int) -> bool:
    if number < 0:
        raise NegativeNumber('Значение не может быть отрицательным')
    return number > 0

try:
    check_number(-5)
except NegativeNumber as e:
    print(f'Ошибка : {e}')
```

# 2) Настройте простое логирование с использованием модуля logging: создайте лог-файл, в который записываются сообщения об успешном выполнении программы и обработке исключений.

```python
import json
import logging


logging.basicConfig(
    filename='app.log', # файл с логами
    filemode='w', # 'w' для перезаписи, 'a' для дополнения записи
    format='%(asctime)s - %(levelname)s - %(message)s',  # формат логов
    encoding='UTF-8',
    level=logging.DEBUG,
    force=True,
)

def check_json(path_json_file):
    
    with open(path_json_file, 'r', encoding='utf-8') as jsonfile:
        data = json.load(jsonfile)
        logging.info('JSON файл успешно загружен')
        
        for user in data['users']:
            if user['active'] is False:
                logging.warning(f'Пользователь {user["username"]} находится в статусе {user["active"]}')

        users_ids = [user['id'] for user in data['users']]

        for order in data['orders']:
            if order['amount'] < 0:
                logging.error(f'Сумма заказа order_id: {order["order_id"]} меньше нуля и равна {order["amount"]}')
            if order['user_id'] not in users_ids:
                logging.error(f'Пользователь с user_id: {order["user_id"]}, у заказа {order["order_id"]} отсутсвует в списке users')
              
        

path_json_file = 'logging_practice.json'
try:
    check_json(path_json_file)
except FileNotFoundError as e:
    logging.error(f'Ошибка при чтении файла, файл не найден: {e}' )
```
